```python
%%writefile /content/src/agcn.py
import torch
import torch.nn as nn
import torch.nn.functional as F


class AGCNBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.conv_t = nn.Conv2d(in_channels, out_channels, kernel_size=(9, 1), padding=(4, 0))
        self.conv_v = nn.Conv2d(out_channels, out_channels, kernel_size=(1, 1))
        self.bn = nn.BatchNorm2d(out_channels)

    def forward(self, x):
        x = self.conv_t(x)
        x = self.conv_v(x)
        x = self.bn(x)
        return F.relu(x)


class AGCNModel(nn.Module):

    def __init__(self, in_channels=3, hidden=64, out_dim=512):
        super().__init__()

        self.data_bn = nn.BatchNorm1d(in_channels * 25)

        self.block1 = AGCNBlock(in_channels, hidden)
        self.block2 = AGCNBlock(hidden, hidden * 2)
        self.block3 = AGCNBlock(hidden * 2, hidden * 4)

        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.fc = nn.Linear(hidden * 4, out_dim)

    def forward(self, x):
        B, T, V, C = x.shape

        x = x.permute(0, 3, 1, 2).contiguous()

        x = x.view(B, C * V, T)
        x = self.data_bn(x)
        x = x.view(B, C, V, T).permute(0, 1, 3, 2)

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)

        x = self.pool(x).view(B, -1)
        x = self.fc(x)
        return x
```

```python
%%writefile /content/src/feature_extract.py
import os
import glob
import hashlib
import sys
from pathlib import Path
from typing import Dict, Optional, Tuple

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F

import tensorflow as tf
from tensorflow import keras

from decord import VideoReader, cpu

try:
    import torch
    TORCH_AVAILABLE = True
except ImportError:
    TORCH_AVAILABLE = False

try:
    import tensorflow as tf
    POSENET_AVAILABLE = True
except ImportError:
    POSENET_AVAILABLE = False


MIGA_ROOT = os.environ.get("MIGA_ROOT", "/content/miga_data")

RGB_P1_TRAIN = f"{MIGA_ROOT}/imigue_rgb_phase1/train_data"
RGB_P1_VAL   = f"{MIGA_ROOT}/imigue_rgb_phase1/validation_data"
RGB_P2       = f"{MIGA_ROOT}/imigue_rgb_phase2"

SK_P1_TRAIN  = f"{MIGA_ROOT}/imigue_data_phase1/datasets/imigue_skeleton_train"
SK_P1_VAL    = f"{MIGA_ROOT}/imigue_data_phase1/datasets/imigue_skeleton_validate"
SK_P2_TEST   = f"{MIGA_ROOT}/imigue_data_phase2/imigue_skeleton_test"


CTX_BACKBONE  = os.environ.get("CTX_BACKBONE", "r3d_18")
FACE_MODEL    = os.environ.get("FACE_MODEL", "emotieff")
FACE_SAMPLING = os.environ.get("FACE_SAMPLING", "mid")
FACE_DETECT   = os.environ.get("FACE_DETECT", "yolo")
SKEL_MODEL    = os.environ.get("SKEL_MODEL", "agcn")
SKEL_AGG      = os.environ.get("SKEL_AGG", "mean")


def vid4(x: int) -> str:
    return f"{int(x):04d}"


def resolve_video_path_phase1(video_id: int, split: str) -> Optional[str]:
    v = vid4(video_id)
    if split == "train":
        p = os.path.join(RGB_P1_TRAIN, v, f"{v}.mp4")
        return p if os.path.exists(p) else None
    if split == "val":
        p = os.path.join(RGB_P1_VAL, v, f"{v}.mp4")
        return p if os.path.exists(p) else None
    for p in [
        os.path.join(RGB_P1_TRAIN, v, f"{v}.mp4"),
        os.path.join(RGB_P1_VAL, v, f"{v}.mp4"),
    ]:
        if os.path.exists(p):
            return p
    return None


def resolve_video_path_phase2(video_id: int) -> Optional[str]:
    v = vid4(video_id)
    p = os.path.join(RGB_P2, v, f"{v}.mp4")
    if os.path.exists(p):
        return p
    hits = glob.glob(os.path.join(RGB_P2, "**", f"{v}.mp4"), recursive=True)
    return hits[0] if hits else None


def resolve_skeleton_path_phase1(video_id: int, split: str, prefer_hand: bool = True) -> Optional[str]:
    v = vid4(video_id)
    base = SK_P1_TRAIN if split == "train" else SK_P1_VAL
    p_hand  = os.path.join(base, v, f"{v}_light_hand.csv")
    p_light = os.path.join(base, v, f"{v}_light.csv")
    if prefer_hand and os.path.exists(p_hand): return p_hand
    if os.path.exists(p_light): return p_light
    if os.path.exists(p_hand):  return p_hand
    return None


def resolve_skeleton_path_phase2(video_id: int, prefer_hand: bool = True) -> Optional[str]:
    v = vid4(video_id)
    p_hand  = os.path.join(SK_P2_TEST, v, f"{v}_light_hand.csv")
    p_light = os.path.join(SK_P2_TEST, v, f"{v}_light.csv")
    if prefer_hand and os.path.exists(p_hand): return p_hand
    if os.path.exists(p_light): return p_light
    if os.path.exists(p_hand):  return p_hand
    return None


def cache_key(video_id: int, split: str, phase: int) -> str:
    s = f"{int(video_id)}|{split}|{phase}"
    return hashlib.md5(s.encode("utf-8")).hexdigest()[:12]


def cache_paths(feat_dir: str, video_id: int, split: str, phase: int) -> Dict[str, str]:
    key = cache_key(video_id, split, phase)
    d = Path(feat_dir)
    d.mkdir(parents=True, exist_ok=True)
    return {
        "ctx":   str(d / f"{key}_ctx.pt"),
        "face":  str(d / f"{key}_face.pt"),
        "fmask": str(d / f"{key}_facemask.pt"),
        "skel":  str(d / f"{key}_skel.pt"),
        "smask": str(d / f"{key}_skelmask.pt"),
    }


def _vr(path: str) -> VideoReader:
    return VideoReader(path, ctx=cpu(0))


# ============================
# CTX: Video backbones
# ============================

_CTX_MODEL = None

def build_ctx_model(device: str = "cuda"):
    global _CTX_MODEL
    if _CTX_MODEL is not None:
        _CTX_MODEL.to(device).eval()
        return _CTX_MODEL

    if CTX_BACKBONE == "mc3_18":
        from torchvision.models.video import mc3_18, MC3_18_Weights
        weights = MC3_18_Weights.DEFAULT
        model = mc3_18(weights=weights)
        model.fc = nn.Identity()
    elif CTX_BACKBONE == "r2plus1d_18":
        from torchvision.models.video import r2plus1d_18, R2Plus1D_18_Weights
        weights = R2Plus1D_18_Weights.DEFAULT
        model = r2plus1d_18(weights=weights)
        model.fc = nn.Identity()
    else:
        from torchvision.models.video import r3d_18, R3D_18_Weights
        weights = R3D_18_Weights.DEFAULT
        model = r3d_18(weights=weights)
        model.fc = nn.Identity()

    model = model.to(device).eval()
    _CTX_MODEL = model
    return _CTX_MODEL


def _preprocess_clip_torch(clip_rgb: np.ndarray, device: str) -> torch.Tensor:
    x = torch.from_numpy(clip_rgb).to(torch.float32) / 255.0
    x = x.permute(3, 0, 1, 2)
    x = x.unsqueeze(0)
    target_size = (112, 112)
    x = F.interpolate(x, size=(x.shape[2], *target_size), mode="trilinear", align_corners=False)

    mean = torch.tensor([0.43216, 0.394666, 0.37645], device=x.device).view(1,3,1,1,1)
    std  = torch.tensor([0.22803, 0.22145, 0.216989], device=x.device).view(1,3,1,1,1)
    x = (x - mean) / std
    return x.to(device)


@torch.no_grad()
def extract_ctx_stream(video_path: str, device: str = "cuda", chunk: int = 32) -> torch.Tensor:
    model = build_ctx_model(device=device)
    vr = _vr(video_path)
    T = len(vr)
    if T <= 0:
        return torch.zeros((0, 512), dtype=torch.float32)

    pad = (-T) % chunk
    total = T + pad

    feats = []
    for i in range(0, total, chunk):
        idxs = list(range(i, min(i + chunk, T)))
        if len(idxs) == 0:
            break
        frames = vr.get_batch(idxs).asnumpy()
        if frames.shape[0] < chunk:
            frames = np.concatenate([frames, np.repeat(frames[-1:], chunk - frames.shape[0], axis=0)], axis=0)

        x = _preprocess_clip_torch(frames, device=device)
        f = model(x).squeeze(0).detach().cpu()
        feats.append(f)

    return torch.stack(feats, dim=0).to(torch.float32)


# ============================
# FACE: Models
# ============================
_FACE_MODEL = None
_EMO_REC = None

def _init_yolo():
    global _FACE_MODEL
    if _FACE_MODEL is not None:
        return _FACE_MODEL
    try:
        from ultralytics import YOLO
        for w in ["yolov8n-face.pt", "yolov8n.pt"]:
            try:
                _FACE_MODEL = YOLO(w)
                break
            except Exception:
                _FACE_MODEL = None
    except Exception:
        _FACE_MODEL = None
    return _FACE_MODEL


def _center_crop(img, size=224):
    h, w = img.shape[:2]
    s = min(h, w)
    y0 = (h - s) // 2
    x0 = (w - s) // 2
    crop = img[y0:y0+s, x0:x0+s]
    import cv2
    crop = cv2.resize(crop, (size, size))
    return crop


def _detect_face_crop(img_rgb, size=224):
    if FACE_DETECT == "center":
        return _center_crop(img_rgb, size=size), True

    model = _init_yolo()
    if model is None:
        return _center_crop(img_rgb, size=size), False

    try:
        res = model.predict(source=img_rgb)
        if len(res) == 0 or res[0].boxes is None or len(res[0].boxes) == 0:
            return _center_crop(img_rgb, size=size), False
        boxes = res[0].boxes.xyxy.detach().cpu().numpy()
        areas = (boxes[:,2]-boxes[:,0])*(boxes[:,3]-boxes[:,1])
        b = boxes[int(np.argmax(areas))]
        x1, y1, x2, y2 = [int(max(0, v)) for v in b]
        x2 = min(x2, img_rgb.shape[1]-1)
        y2 = min(y2, img_rgb.shape[0]-1)
        crop = img_rgb[y1:y2, x1:x2]
        if crop.size == 0:
            return _center_crop(img_rgb, size=size), False
        import cv2
        crop = cv2.resize(crop, (size, size))
        return crop, True
    except Exception:
        return _center_crop(img_rgb, size=size), False


def _init_face_embedding_model(device: str):
    global _EMO_REC
    if _EMO_REC is not None:
        return _EMO_REC


    if FACE_MODEL == "emotieff":
        from emotiefflib.facial_analysis import EmotiEffLibRecognizerTorch
        try:
            _EMO_REC = EmotiEffLibRecognizerTorch(device=device)
        except TypeError:
            _EMO_REC = EmotiEffLibRecognizerTorch()
        return _EMO_REC

    elif FACE_MODEL == "facenet_vggface2":
        from facenet_pytorch import InceptionResnetV1
        _EMO_REC = InceptionResnetV1(pretrained='vggface2').eval().to(device)
        return _EMO_REC

    elif FACE_MODEL == "facenet_casia":
        from facenet_pytorch import InceptionResnetV1
        _EMO_REC = InceptionResnetV1(pretrained='casia-webface').eval().to(device)
        return _EMO_REC

    else:
        raise ValueError(f"Unknown FACE_MODEL: {FACE_MODEL}")

@torch.no_grad()
def extract_face_windows_mtcnn(video_path: str, device: str = "cuda", chunk: int = 32) -> Tuple[torch.Tensor, torch.Tensor]:
    vr = _vr(video_path)
    T = len(vr)
    if T <= 0:
        D = 512
        return torch.zeros((0, D), dtype=torch.float32), torch.zeros((0,), dtype=torch.bool)

    pad = (-T) % chunk
    total = T + pad

    feats = []
    mask = []
    D = None

    for i in range(0, total, chunk):
        if FACE_SAMPLING == "3frames_avg":
            idxs = [
                min(i + max(0, chunk // 4), T - 1),
                min(i + max(0, chunk // 2), T - 1),
                min(i + max(0, (3 * chunk) // 4), T - 1),
            ]
        else:
            idxs = [min(i + chunk // 2, T - 1)]

        embs = []
        oks = []
        for idx in idxs:
            frame = vr[idx].asnumpy()
            crop, ok = _detect_face_crop(frame, size=224)
            if ok:
                emb = emotieff_embed(crop, device=device)
                embs.append(emb)
                oks.append(True)
            else:
                oks.append(False)

        if any(oks):
            emb = np.mean(embs, axis=0).astype(np.float32)
            if D is None:
                D = int(emb.shape[0])
            feats.append(torch.from_numpy(emb).to(torch.float32))
            mask.append(True)
        else:
            if D is None:
                D = 512
            feats.append(torch.zeros((D,), dtype=torch.float32))
            mask.append(False)

    face = torch.stack(feats, dim=0)
    fmask = torch.tensor(mask, dtype=torch.bool)
    return face, fmask

def _as_rgb_uint8(face_rgb_224):
    x = np.asarray(face_rgb_224)
    if x.dtype != np.uint8:
        x = np.clip(x, 0, 255).astype(np.uint8)
    return x


@torch.no_grad()
def emotieff_embed(face_rgb_224: np.ndarray, device: str = "cuda") -> np.ndarray:
    rec = _init_face_embedding_model(device)
    x = _as_rgb_uint8(face_rgb_224)

    if FACE_MODEL == "emotieff":
        for name in ["get_face_embedding","get_embedding","get_embeddings","extract_embedding","extract_embeddings","get_features","extract_features"]:
            if hasattr(rec, name) and callable(getattr(rec, name)):
                out = getattr(rec, name)(x)
                return np.asarray(out).reshape(-1).astype(np.float32)
        if callable(rec):
            out = rec(x)
            if isinstance(out, dict):
                for k in ["embedding","emb","features","feature"]:
                    if k in out:
                        return np.asarray(out[k]).reshape(-1).astype(np.float32)
            return np.asarray(out).reshape(-1).astype(np.float32)
        raise RuntimeError("EmotiEffLib: не удалось получить embedding")

    elif FACE_MODEL in ["facenet_vggface2", "facenet_casia"]:
        import cv2
        x_resized = cv2.resize(x, (160, 160))
        img = torch.from_numpy(x_resized).permute(2,0,1).unsqueeze(0).float().to(device)
        img = (img - 127.5) / 128.0
        emb = rec(img).cpu().numpy().flatten().astype(np.float32)
        return emb

    else:
        raise RuntimeError(f"Не поддерживается модель {FACE_MODEL}")


@torch.no_grad()
def extract_face_windows(video_path: str, device: str = "cuda", chunk: int = 32) -> Tuple[torch.Tensor, torch.Tensor]:
    vr = _vr(video_path)
    T = len(vr)
    if T <= 0:
        if FACE_MODEL == "emotieff":
            D = 1280
        elif FACE_MODEL in ["facenet_vggface2", "facenet_casia"]:
            D = 512
        else:
            D = 512
        return torch.zeros((0, D), dtype=torch.float32), torch.zeros((0,), dtype=torch.bool)

    pad = (-T) % chunk
    total = T + pad

    feats = []
    mask = []
    D = None

    for i in range(0, total, chunk):
        if FACE_SAMPLING == "3frames_avg":
            idxs = [
                min(i + max(0, chunk // 4), T - 1),
                min(i + max(0, chunk // 2), T - 1),
                min(i + max(0, (3 * chunk) // 4), T - 1),
            ]
        else:
            idxs = [min(i + chunk // 2, T - 1)]

        embs = []
        oks = []
        for idx in idxs:
            frame = vr[idx].asnumpy()
            crop, ok = _detect_face_crop(frame, size=224)
            if ok:
                emb = emotieff_embed(crop, device=device)
                embs.append(emb)
                oks.append(True)
            else:
                oks.append(False)

        if any(oks):
            emb = np.mean(embs, axis=0).astype(np.float32)
            if D is None:
                D = int(emb.shape[0])
            feats.append(torch.from_numpy(emb).to(torch.float32))
            mask.append(True)
        else:
            if D is None:
                if FACE_MODEL == "emotieff":
                    D = 1280
                elif FACE_MODEL in ["facenet_vggface2", "facenet_casia"]:
                    D = 512
                else:
                    D = 512
            feats.append(torch.zeros((D,), dtype=torch.float32))
            mask.append(False)

    face = torch.stack(feats, dim=0)
    fmask = torch.tensor(mask, dtype=torch.bool)
    return face, fmask


# ============================
# SKEL: Models (AGCN)
# ============================
_SKEL_PROJ = None
_SKEL_MODEL_OBJ = None

def build_skel_model(device: str = "cpu"):
    global _SKEL_MODEL_OBJ
    if _SKEL_MODEL_OBJ is not None:
        return _SKEL_MODEL_OBJ

    if SKEL_MODEL == "agcn":
        from .agcn import AGCNModel
        _SKEL_MODEL_OBJ = AGCNModel(in_channels=3, hidden=64, out_dim=512).to(device).eval()
        return _SKEL_MODEL_OBJ
    else:
        raise NotImplementedError(f"Модель {SKEL_MODEL} не реализована.")


def _skel_projector(in_dim: int, out_dim: int = 512) -> nn.Module:
    global _SKEL_PROJ
    if _SKEL_PROJ is not None and getattr(_SKEL_PROJ, "_in_dim", None) == in_dim:
        return _SKEL_PROJ
    torch.manual_seed(42)
    proj = nn.Sequential(
        nn.Linear(in_dim, out_dim),
        nn.LayerNorm(out_dim),
        nn.GELU(),
    )
    proj._in_dim = in_dim
    _SKEL_PROJ = proj.eval()
    return _SKEL_PROJ


def _load_skeleton_csv(path: str) -> np.ndarray:
    df = pd.read_csv(path)
    return df.values.astype(np.float32)


@torch.no_grad()
def extract_skeleton_windows(skel_csv_path: str, device: str = "cpu", chunk: int = 32) -> Tuple[torch.Tensor, torch.Tensor]:
    if SKEL_MODEL == "agcn":
        arr = _load_skeleton_csv(skel_csv_path)
        T, D = arr.shape
        if T <= 0:
            return torch.zeros((0, 512), dtype=torch.float32), torch.zeros((0,), dtype=torch.bool)

        pad = (-T) % chunk
        total = T + pad
        if pad:
            arr = np.concatenate([arr, np.repeat(arr[-1:], pad, axis=0)], axis=0)

        if SKEL_AGG == "mean_std":
            in_dim = 2 * D
        else:
            in_dim = D

        proj = _skel_projector(in_dim, 512)

        feats = []
        mask = []
        for i in range(0, total, chunk):
            w = arr[i:i+chunk]
            valid = not np.allclose(w, 0)

            if SKEL_AGG == "max":
                v = w.max(axis=0)
            elif SKEL_AGG == "mean_std":
                mu = w.mean(axis=0)
                sd = w.std(axis=0)
                v = np.concatenate([mu, sd], axis=0)
            else:
                v = w.mean(axis=0)

            x = torch.from_numpy(v).to(torch.float32)
            z = proj(x).detach().cpu()
            feats.append(z)
            mask.append(valid)

        skel = torch.stack(feats, dim=0).to(torch.float32)
        smask = torch.tensor(mask, dtype=torch.bool)
        return skel, smask

    else:
        raise NotImplementedError(f"Модель {SKEL_MODEL} не поддерживается.")


def build_skel_model(device: str = "cpu"):
    global _SKEL_MODEL_OBJ
    if _SKEL_MODEL_OBJ is not None:
        return _SKEL_MODEL_OBJ

    if SKEL_MODEL == "gcn":
        _SKEL_MODEL_OBJ = GCNModel(in_channels=3, hidden=64, out_dim=512).to(device).eval()
        return _SKEL_MODEL_OBJ
    elif SKEL_MODEL == "mlp":
        _SKEL_MODEL_OBJ = MLPModel(in_channels=3, hidden=64, out_dim=512).to(device).eval()
        return _SKEL_MODEL_OBJ
    else:
        raise NotImplementedError(f"Модель {SKEL_MODEL} не реализована.")

def load_posenet_model():
    model = tf.saved_model.load("https://tfhub.dev/tensorflow/posenet/mobilenet/4")
    return model


def build_skel_model(device: str = "cpu"):
    global _SKEL_MODEL_OBJ
    if _SKEL_MODEL_OBJ is not None:
        return _SKEL_MODEL_OBJ

    if SKEL_MODEL == "posenet" and POSENET_AVAILABLE:
        _SKEL_MODEL_OBJ = load_posenet_model()
        return _SKEL_MODEL_OBJ
    else:
        raise NotImplementedError(f"Модель {SKEL_MODEL} не реализована.")

def load_hrnet_model():
    config_file = 'configs/top_down/hrnet/hrnet_w32_coco_256x192.py'
    checkpoint_file = 'checkpoints/hrnet_w32_coco_256x192-ef1e5e1e.pth'
    pose_model = init_pose_model(config_file, checkpoint_file, device='cuda')
    return pose_model


def build_skel_model(device: str = "cpu"):
    global _SKEL_MODEL_OBJ
    if _SKEL_MODEL_OBJ is not None:
        return _SKEL_MODEL_OBJ

    if SKEL_MODEL == "hrnet":
        _SKEL_MODEL_OBJ = load_hrnet_model()
        return _SKEL_MODEL_OBJ
    else:
        raise NotImplementedError(f"Модель {SKEL_MODEL} не реализована.")


def extract_skeleton_windows(skel_csv_path: str, device: str = "cpu", chunk: int = 32) -> Tuple[torch.Tensor, torch.Tensor]:
    arr = _load_skeleton_csv(skel_csv_path)
    T, D = arr.shape
    if T <= 0:
        return torch.zeros((0, 512), dtype=torch.float32), torch.zeros((0,), dtype=torch.bool)

    pad = (-T) % chunk
    total = T + pad
    if pad:
        arr = np.concatenate([arr, np.repeat(arr[-1:], pad, axis=0)], axis=0)

    proj = _skel_projector(D, 512)

    feats = []
    mask = []
    for i in range(0, total, chunk):
        w = arr[i:i+chunk]
        valid = not np.allclose(w, 0)
        v = w.mean(axis=0)

        x = torch.from_numpy(v).to(torch.float32)
        z = proj(x).detach().cpu()
        feats.append(z)
        mask.append(valid)

    skel = torch.stack(feats, dim=0).to(torch.float32)
    smask = torch.tensor(mask, dtype=torch.bool)
    return skel, smask

@torch.no_grad()
def build_and_cache_one(
    video_id: int,
    split: str,
    phase: int,
    feat_dir: str,
    chunk: int = 32,
    vpath: Optional[str] = None,
    spath: Optional[str] = None,
    do_ctx: bool = True,
    do_face: bool = True,
    do_skel: bool = True,
    device: str = "cuda",
):
    ps = cache_paths(feat_dir, video_id, split, phase)

    if vpath is None:
        vpath = resolve_video_path_phase1(video_id, split) if phase == 1 else resolve_video_path_phase2(video_id)

    if spath is None:
        spath = resolve_skeleton_path_phase1(video_id, split) if phase == 1 else resolve_skeleton_path_phase2(video_id)

    if do_ctx or do_face:
        if vpath is None or (not os.path.exists(vpath)):
            raise FileNotFoundError(f"RGB video not found for id={video_id} split={split} phase={phase}")

    if do_skel:
        if spath is None or (not os.path.exists(spath)):
            raise FileNotFoundError(f"Skeleton csv not found for id={video_id} split={split} phase={phase}")

    if do_ctx and (not os.path.exists(ps["ctx"])):
        ctx = extract_ctx_stream(vpath, device=device, chunk=chunk)
        torch.save(ctx, ps["ctx"])

    if do_face and (not os.path.exists(ps["face"])) and (not os.path.exists(ps["fmask"])):
        face, fmask = extract_face_windows(vpath, device=device, chunk=chunk)
        torch.save(face, ps["face"])
        torch.save(fmask, ps["fmask"])

    if do_skel and (not os.path.exists(ps["skel"])) and (not os.path.exists(ps["smask"])):
        skel, smask = extract_skeleton_windows(spath, device="cpu", chunk=chunk)
        torch.save(skel, ps["skel"])
        torch.save(smask, ps["smask"])

    return ps
```

```python
%%writefile /content/src/cache_features_cli.py
import argparse
import pandas as pd
from tqdm import tqdm
import torch
import traceback

from src.feature_extract import build_and_cache_one

def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--csv", required=True)
    ap.add_argument("--phase", type=int, required=True)
    ap.add_argument("--feat_dir", required=True)
    ap.add_argument("--chunk", type=int, default=32)
    ap.add_argument("--start", type=int, default=0)
    ap.add_argument("--end", type=int, default=999999)
    ap.add_argument("--split_col", type=str, default="split")
    ap.add_argument("--device", choices=["auto","cpu","cuda"], default="auto")
    ap.add_argument("--do_ctx", action="store_true")
    ap.add_argument("--do_face", action="store_true")
    ap.add_argument("--do_skel", action="store_true")
    args = ap.parse_args()

    if args.device == "auto":
        device = "cuda" if torch.cuda.is_available() else "cpu"
    else:
        device = args.device

    if not (args.do_ctx or args.do_face or args.do_skel):
        args.do_ctx = args.do_face = args.do_skel = True

    print("DEVICE:", device)
    print("FEAT_DIR:", args.feat_dir)
    print("MODES: ctx=", args.do_ctx, "face=", args.do_face, "skel=", args.do_skel)

    df = pd.read_csv(args.csv)
    df = df.iloc[args.start:args.end].reset_index(drop=True)

    ok, bad = 0, 0

    for i, r in tqdm(df.iterrows(), total=len(df)):
        try:
            vid = int(r["video_id"])
            split = r.get(args.split_col, "test")

            build_and_cache_one(
                video_id=vid,
                split=split,
                phase=args.phase,
                feat_dir=args.feat_dir,
                chunk=args.chunk,
                vpath=None,
                spath=r.get("skeleton_path", None),
                do_ctx=args.do_ctx,
                do_face=args.do_face,
                do_skel=args.do_skel,
                device=device,
            )
            ok += 1

        except Exception:
            bad += 1
            print(" ERROR")
            print("row:", i, "video_id:", r.get("video_id"), "split:", r.get(args.split_col, ""))
            if "skeleton_path" in r:
                print("skeleton_path:", r.get("skeleton_path"))
            traceback.print_exc()

    print(f"\nDONE shard [{args.start}:{args.end}) ok={ok} bad={bad}")

if __name__ == "__main__":
    main()
```

```python
%%writefile /content/src/model.py
import torch
import torch.nn as nn
import torch.nn.functional as F

class AttnPool(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.scorer = nn.Sequential(
            nn.Linear(d, d // 2),
            nn.GELU(),
            nn.Linear(d // 2, 1)
        )

    def forward(self, x, mask=None):
        """
        x: (batch, seq_len, d)
        mask: (batch, seq_len) булева маска, True для валидных токенов
        """
        if mask is None:
            return x.mean(dim=1)

        if mask.numel() == 0 or mask.sum() == 0:
            return torch.zeros(x.size(0), x.size(2), device=x.device, dtype=x.dtype)

        score = self.scorer(x).squeeze(-1)
        score = score.float()
        score = score.masked_fill(~mask, -1e4)
        w = torch.softmax(score, dim=1).unsqueeze(-1)
        w = w.to(dtype=x.dtype)
        return (x * w).sum(dim=1)

class ContextModel(nn.Module):
    def __init__(self, d_ctx_in, d=256, n_layers=4, n_heads=4, dropout=0.2):
        super().__init__()
        self.ctx_in = nn.Sequential(nn.Linear(d_ctx_in, d), nn.Dropout(dropout))
        enc = nn.TransformerEncoderLayer(d_model=d, nhead=n_heads, dim_feedforward=4*d,
                                         dropout=dropout, batch_first=True, activation="gelu")
        self.ctx_enc = nn.TransformerEncoder(enc, num_layers=n_layers)
        self.ctx_pool = AttnPool(d)
        self.head = nn.Sequential(nn.Dropout(dropout), nn.Linear(d, 1))

    def forward(self, ctx):
        x_ctx = self.ctx_in(ctx)
        x_ctx = self.ctx_enc(x_ctx)
        ctx_vec = self.ctx_pool(x_ctx)
        return self.head(ctx_vec).squeeze(-1)

class FaceModel(nn.Module):
    def __init__(self, d_face_in, d=256, n_layers=4, n_heads=4, dropout=0.2):
        super().__init__()
        self.face_in = nn.Sequential(nn.Linear(d_face_in, d), nn.Dropout(dropout))
        enc = nn.TransformerEncoderLayer(d_model=d, nhead=n_heads, dim_feedforward=4*d,
                                         dropout=dropout, batch_first=True, activation="gelu")
        self.face_enc = nn.TransformerEncoder(enc, num_layers=n_layers)
        self.face_pool = AttnPool(d)
        self.head = nn.Sequential(nn.Dropout(dropout), nn.Linear(d, 1))

    def forward(self, face):
        x_face = self.face_in(face)
        x_face = self.face_enc(x_face)
        face_vec = self.face_pool(x_face)
        return self.head(face_vec).squeeze(-1)

class SkelModel(nn.Module):
    def __init__(self, d_skel_in, d=256, n_layers=4, n_heads=4, dropout=0.2):
        super().__init__()
        self.skel_in = nn.Sequential(nn.Linear(d_skel_in, d), nn.Dropout(dropout))
        enc = nn.TransformerEncoderLayer(d_model=d, nhead=n_heads, dim_feedforward=4*d,
                                         dropout=dropout, batch_first=True, activation="gelu")
        self.skel_enc = nn.TransformerEncoder(enc, num_layers=n_layers)
        self.skel_pool = AttnPool(d)
        self.head = nn.Sequential(nn.Dropout(dropout), nn.Linear(d, 1))

    def forward(self, skel):
        x_skel = self.skel_in(skel)
        x_skel = self.skel_enc(x_skel)
        skel_vec = self.skel_pool(x_skel)
        return self.head(skel_vec).squeeze(-1)

class FocalLoss(nn.Module):
    def __init__(self, alpha=None, gamma=2.0, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction

    def forward(self, inputs, targets):
        BCE_loss = nn.functional.binary_cross_entropy_with_logits(inputs, targets, reduction='none')
        pt = torch.exp(-BCE_loss)
        focal_loss = (1 - pt) ** self.gamma * BCE_loss
        if self.alpha is not None:
            alpha_t = self.alpha[targets.long()]
            focal_loss = alpha_t * focal_loss
        if self.reduction == 'mean':
            return focal_loss.mean()
        elif self.reduction == 'sum':
            return focal_loss.sum()
        return focal_loss

def create_optimizer(model, lr=1e-5):
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-6)
    return optimizer
```

```python
%%writefile /content/src/ds.py
import os
import torch
import numpy as np
from torch.utils.data import Dataset, DataLoader
from torch.utils.data import WeightedRandomSampler

def cache_key(video_id, split, phase):
    s = f"{int(video_id)}|{split}|{phase}"
    import hashlib
    return hashlib.md5(s.encode("utf-8")).hexdigest()[:12]

class Track3CachedDataset(Dataset):
    def __init__(self, df, ctx_dir=None, face_dir=None, skel_dir=None, phase=1, has_label=True, use_ctx=True, use_face=True, use_skel=True, ctx_dim=512, face_dim=1280, skel_dim=512):
        self.df = df.reset_index(drop=True)
        self.phase = phase
        self.has_label = has_label
        self.use_ctx = use_ctx
        self.use_face = use_face
        self.use_skel = use_skel
        self.ctx_dim = ctx_dim
        self.face_dim = face_dim
        self.skel_dim = skel_dim

        self.ctx_dir = ctx_dir
        self.face_dir = face_dir
        self.skel_dir = skel_dir

        self.keys = []
        self.ids = []
        self.splits = []
        for _, row in self.df.iterrows():
            video_id = int(row["video_id"])
            split = row.get("split", "train")
            key = cache_key(video_id, split, self.phase)
            self.keys.append(key)
            self.ids.append(video_id)
            self.splits.append(split)

        if has_label and "label" in self.df.columns:
            self.labels = self.df["label"].values.astype(np.float32)
        else:
            self.labels = None

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        key = self.keys[idx]
        out = {"id": self.ids[idx], "split": self.splits[idx]}

        if self.use_ctx and self.ctx_dir:
            ctx_path = os.path.join(self.ctx_dir, f"{key}_ctx.pt")
            if os.path.exists(ctx_path):
                ctx = torch.load(ctx_path)
                out["ctx"] = ctx.float()
            else:
                out["ctx"] = torch.zeros((1, self.ctx_dim), dtype=torch.float32)
        else:
            out["ctx"] = torch.zeros((1, self.ctx_dim), dtype=torch.float32)

        if self.use_face and self.face_dir:
            face_path = os.path.join(self.face_dir, f"{key}_face.pt")
            fmask_path = os.path.join(self.face_dir, f"{key}_facemask.pt")
            if os.path.exists(face_path) and os.path.exists(fmask_path):
                face = torch.load(face_path)
                fmask = torch.load(fmask_path)
                out["face"] = face.float()
                out["face_mask"] = fmask.bool()
            else:
                out["face"] = torch.zeros((1, self.face_dim), dtype=torch.float32)
                out["face_mask"] = torch.zeros(1, dtype=torch.bool)
        else:
            out["face"] = torch.zeros((1, self.face_dim), dtype=torch.float32)
            out["face_mask"] = torch.zeros(1, dtype=torch.bool)

        if self.use_skel and self.skel_dir:
            skel_path = os.path.join(self.skel_dir, f"{key}_skel.pt")
            smask_path = os.path.join(self.skel_dir, f"{key}_skelmask.pt")
            if os.path.exists(skel_path) and os.path.exists(smask_path):
                skel = torch.load(skel_path)
                smask = torch.load(smask_path)
                out["skel"] = skel.float()
                out["skel_mask"] = smask.bool()
            else:
                out["skel"] = torch.zeros((1, self.skel_dim), dtype=torch.float32)
                out["skel_mask"] = torch.zeros(1, dtype=torch.bool)
        else:
            out["skel"] = torch.zeros((1, self.skel_dim), dtype=torch.float32)
            out["skel_mask"] = torch.zeros(1, dtype=torch.bool)

        if self.labels is not None:
            out["y"] = torch.tensor(self.labels[idx], dtype=torch.float32)

        return out
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from sklearn.metrics import accuracy_score, roc_auc_score
import os, hashlib
import pandas as pd
import numpy as np
from tqdm import tqdm
from src.ds import Track3CachedDataset
from src.model import ContextModel, FaceModel, SkelModel, FocalLoss

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Устройство: {device}")

BASE = "/content/drive/MyDrive/miga_features_cache_agcn"

BATCH_SIZE  = 8
LR          = 1e-6
NUM_EPOCHS  = 20
HIDDEN_DIM  = 512
N_LAYERS    = 8
N_HEADS     = 4
DROPOUT     = 0.5
FOCAL_GAMMA = 0.5
PATIENCE    = 7

experiments = [
    {"name": "ResNet-18",   "modality": "ctx",  "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "MC3-18",      "modality": "ctx",  "ctx_dir": f"{BASE}/exp_ctx_mc3",       "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "R2Plus1D-18", "modality": "ctx",  "ctx_dir": f"{BASE}/exp_ctx_r2plus1d",  "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "YOLO8",       "modality": "face", "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "FaceNet",     "modality": "face", "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/exp_face_facenet",  "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "EmotiEffLib", "modality": "face", "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/exp_face_emotieff", "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "2s-AGCN",     "modality": "skel", "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "HRNet",       "modality": "skel", "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_hrnet"},
    {"name": "PoseNet",     "modality": "skel", "ctx_dir": f"{BASE}/baseline_ctx",      "face_dir": f"{BASE}/baseline_face",     "skel_dir": f"{BASE}/exp_skel_posenet"},
]

def make_cache_key(video_id, split, phase=1):
    return hashlib.md5(f"{int(video_id)}|{split}|{phase}".encode()).hexdigest()[:12]

def fast_filter_by_consistent_dim(df, feat_dir, suffix, phase=1):
    dims, missing = {}, 0
    for _, row in df.iterrows():
        path = os.path.join(feat_dir, f"{make_cache_key(row['video_id'], row['split'], phase)}_{suffix}.pt")
        if not os.path.exists(path): missing += 1; continue
        try:
            d = torch.load(path, map_location="cpu").shape[-1]
            dims[d] = dims.get(d, 0) + 1
        except: missing += 1
    if not dims:
        raise ValueError(f"Ни одного файла: {feat_dir}, suffix={suffix}")
    if missing:
        print(f"   Не найдено {missing}/{len(df)} файлов")
    best_dim = max(dims, key=dims.get)
    valid_idx = []
    for i, (_, row) in enumerate(df.iterrows()):
        path = os.path.join(feat_dir, f"{make_cache_key(row['video_id'], row['split'], phase)}_{suffix}.pt")
        if not os.path.exists(path): continue
        try:
            if torch.load(path, map_location="cpu").shape[-1] == best_dim:
                valid_idx.append(i)
        except: continue
    return df.iloc[valid_idx].reset_index(drop=True), best_dim

def collate_fn(batch):
    out = {k: torch.nn.utils.rnn.pad_sequence(
               [item[k] for item in batch], batch_first=True, padding_value=0)
           for k in ["ctx", "face", "skel"]}
    out["y"] = torch.tensor([item["y"] for item in batch], dtype=torch.float32)
    return out

def initialize_model(modality, feat_dim):
    kwargs = dict(d=HIDDEN_DIM, n_layers=N_LAYERS, n_heads=N_HEADS, dropout=DROPOUT)
    if modality == "ctx":   return ContextModel(d_ctx_in=feat_dim,  **kwargs).to(device)
    elif modality == "face": return FaceModel(d_face_in=feat_dim,   **kwargs).to(device)
    elif modality == "skel": return SkelModel(d_skel_in=feat_dim,   **kwargs).to(device)

class EarlyStopping:
    def __init__(self, patience=PATIENCE):
        self.patience = patience; self.best = None
        self.counter  = 0;        self.early_stop = False
    def __call__(self, score):
        if self.best is None or score > self.best + 1e-4:
            self.best = score; self.counter = 0
        else:
            self.counter += 1
            if self.counter >= self.patience: self.early_stop = True

def train_one_experiment(model, train_loader, valid_loader, criterion, optimizer, modality):
    scheduler      = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="max", factor=0.5, patience=3)
    early_stopping = EarlyStopping(patience=PATIENCE)

    best_state      = None
    best_val_top1   = 0.0
    best_train_top1 = 0.0

    def get_input(batch):
        return batch[modality].to(device)

    for epoch in range(NUM_EPOCHS):
        model.train()
        total_loss, n_correct, n_total = 0.0, 0, 0
        for batch in tqdm(train_loader, desc=f"Ep {epoch+1:02d} train", leave=False):
            inp = get_input(batch)
            y   = batch["y"].to(device)
            optimizer.zero_grad()
            logits = model(inp)
            loss   = criterion(logits, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            total_loss += loss.item()
            preds = (torch.sigmoid(logits) >= 0.5).float()
            n_correct += (preds == y).sum().item()
            n_total   += y.size(0)

        train_top1 = n_correct / n_total

        model.eval()
        probs_list, y_list = [], []
        with torch.no_grad():
            for batch in valid_loader:
                probs_list.extend(torch.sigmoid(model(get_input(batch))).cpu().numpy())
                y_list.extend(batch["y"].numpy())

        probs = np.array(probs_list)

        best_thresh, best_top1_thresh = 0.5, 0.0
        for thresh in np.arange(0.05, 0.95, 0.05):
            acc_t = accuracy_score(y_list, (probs >= thresh).astype(float))
            if acc_t > best_top1_thresh:
                best_top1_thresh = acc_t; best_thresh = thresh
        val_top1 = best_top1_thresh

        try:    val_auc = roc_auc_score(y_list, probs)
        except: val_auc = float("nan")

        print(f"  Ep {epoch+1:02d} | loss={total_loss/len(train_loader):.4f} "
              f"| train_top1={train_top1:.4f} "
              f"| val_top1={val_top1:.4f} "
              f"| thresh={best_thresh:.2f} "
              f"| val_auc={val_auc:.4f} "
              f"| probs=[{probs.min():.3f}..{probs.max():.3f}]")

        scheduler.step(val_top1)

        if val_top1 > best_val_top1:
            best_val_top1   = val_top1
            best_train_top1 = train_top1
            best_state      = {k: v.clone() for k, v in model.state_dict().items()}

        early_stopping(val_top1)
        if early_stopping.early_stop:
            print("  → Early stopping")
            break

    return best_state, best_train_top1, best_val_top1


def run_experiments_cv():
    modality_to_dir = {"ctx": "ctx_dir", "face": "face_dir", "skel": "skel_dir"}
    results = []

    for exp in experiments:
        print(f"\n{'='*65}")
        print(f"  Эксперимент : {exp['name']}  |  Модальность : {exp['modality']}")

        mod      = exp["modality"]
        feat_dir = exp[modality_to_dir[mod]]

        fold_val_top1s = []
        fold_val_aucs  = []

        for fold_idx, (train_idx, val_idx) in enumerate(cv_splits):
            print(f"\n  --- Fold {fold_idx + 1}/{N_FOLDS} ---")

            df_tr_fold = df_all_labeled.iloc[train_idx].reset_index(drop=True)
            df_vl_fold = df_all_labeled.iloc[val_idx].reset_index(drop=True)

            try:
                df_tr, feat_dim = fast_filter_by_consistent_dim(df_tr_fold, feat_dir, mod)
                df_vl, _        = fast_filter_by_consistent_dim(df_vl_fold, feat_dir, mod)
            except ValueError as e:
                print(f"  ПРОПУСК фолда {fold_idx}: {e}"); continue

            if len(df_tr) == 0 or len(df_vl) == 0:
                print(f"  ПРОПУСК фолда {fold_idx}: train={len(df_tr)}, val={len(df_vl)}"); continue

            print(f"  train={len(df_tr)} | val={len(df_vl)} | feat_dim={feat_dim}")

            labels_arr = df_tr["label"].values.astype(int)
            counts     = np.bincount(labels_arr)
            class_w    = torch.tensor([1.0/counts[0], 1.0/counts[1]], dtype=torch.float32).to(device)
            criterion  = FocalLoss(alpha=class_w, gamma=FOCAL_GAMMA)

            sample_w = np.where(labels_arr == 1, 1.0/counts[1], 1.0/counts[0])
            sampler  = torch.utils.data.WeightedRandomSampler(sample_w, len(sample_w), replacement=True)

            train_ds = Track3CachedDataset(df_tr, exp["ctx_dir"], exp["face_dir"], exp["skel_dir"], phase=1)
            valid_ds = Track3CachedDataset(df_vl, exp["ctx_dir"], exp["face_dir"], exp["skel_dir"], phase=1)
            train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler,
                                  collate_fn=collate_fn, drop_last=True)
            valid_dl = DataLoader(valid_ds, batch_size=BATCH_SIZE, shuffle=False,
                                  collate_fn=collate_fn)

            model     = initialize_model(mod, feat_dim)
            optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)

            best_state, best_train_top1, best_val_top1 = train_one_experiment(
                model, train_dl, valid_dl, criterion, optimizer, modality=mod
            )

            save_path = f"/content/{exp['name']}_fold{fold_idx}.pth"
            if best_state is not None:
                torch.save(best_state, save_path)

            if best_state is not None:
                model.load_state_dict(best_state)
            model.eval()
            probs_list, y_list = [], []
            with torch.no_grad():
                for batch in valid_dl:
                    probs_list.extend(
                        torch.sigmoid(model(batch[mod].to(device))).cpu().numpy()
                    )
                    y_list.extend(batch["y"].numpy())
            probs = np.array(probs_list)
            try:    fold_auc = roc_auc_score(y_list, probs)
            except: fold_auc = float("nan")

            fold_val_top1s.append(best_val_top1)
            fold_val_aucs.append(fold_auc)
            print(f"  Fold {fold_idx} → val_top1={best_val_top1:.4f} | val_auc={fold_auc:.4f}")

        if not fold_val_top1s:
            print(f"  Нет результатов для {exp['name']}, пропускаем")
            continue

        mean_top1 = np.mean(fold_val_top1s)
        std_top1  = np.std(fold_val_top1s)
        mean_auc  = np.mean([x for x in fold_val_aucs if not np.isnan(x)])

        print(f"\n  {exp['name']} CV Summary: "
              f"Top-1 = {mean_top1:.4f} ± {std_top1:.4f} | "
              f"AUC = {mean_auc:.4f}")

        results.append({
            "Experiment"       : exp["name"],
            "Modality"         : mod,
            "CV Top-1 Mean"    : round(mean_top1, 4),
            "CV Top-1 Std"     : round(std_top1,  4),
            "CV AUC Mean"      : round(mean_auc,  4),
            "Per-fold Top-1"   : str([round(x, 4) for x in fold_val_top1s]),
        })

    results_df = pd.DataFrame(results)
    print("\n" + "="*65)
    print(f"ИТОГОВЫЕ РЕЗУЛЬТАТЫ — {N_FOLDS}-fold CV")
    print("="*65)
    print(results_df[["Experiment","Modality","CV Top-1 Mean","CV Top-1 Std","CV AUC Mean"]].to_string(index=False))
    return results_df


results_df = run_experiments_cv()
```

```python
import torch
from torch.utils.data import DataLoader
from sklearn.metrics import accuracy_score, roc_auc_score
from scipy.stats import rankdata
import os, hashlib, itertools
import pandas as pd
import numpy as np
from src.ds import Track3CachedDataset
from src.model import ContextModel, FaceModel, SkelModel

device = "cuda" if torch.cuda.is_available() else "cpu"

BASE       = "/content/drive/MyDrive/miga_features_cache_agcn"
HIDDEN_DIM = 512; N_LAYERS = 8; N_HEADS = 4; DROPOUT = 0.5

all_model_configs = [
    {"name": "ResNet-18",   "modality": "ctx",
     "feat_dir": f"{BASE}/baseline_ctx",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "MC3-18",      "modality": "ctx",
     "feat_dir": f"{BASE}/exp_ctx_mc3",
     "ctx_dir":  f"{BASE}/exp_ctx_mc3",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "R2Plus1D-18", "modality": "ctx",
     "feat_dir": f"{BASE}/exp_ctx_r2plus1d",
     "ctx_dir":  f"{BASE}/exp_ctx_r2plus1d",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "YOLO8",       "modality": "face",
     "feat_dir": f"{BASE}/baseline_face",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "FaceNet",     "modality": "face",
     "feat_dir": f"{BASE}/exp_face_facenet",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/exp_face_facenet",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "EmotiEffLib", "modality": "face",
     "feat_dir": f"{BASE}/exp_face_emotieff",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/exp_face_emotieff",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "2s-AGCN",     "modality": "skel",
     "feat_dir": f"{BASE}/exp_skel_mean_std",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_mean_std"},
    {"name": "HRNet",       "modality": "skel",
     "feat_dir": f"{BASE}/exp_skel_hrnet",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_hrnet"},
    {"name": "PoseNet",     "modality": "skel",
     "feat_dir": f"{BASE}/exp_skel_posenet",
     "ctx_dir":  f"{BASE}/baseline_ctx",
     "face_dir": f"{BASE}/baseline_face",
     "skel_dir": f"{BASE}/exp_skel_posenet"},
]


def make_key(vid, split, phase=1):
    return hashlib.md5(f"{int(vid)}|{split}|{phase}".encode()).hexdigest()[:12]

def collate_fn(batch):
    out = {k: torch.nn.utils.rnn.pad_sequence(
               [b[k] for b in batch], batch_first=True, padding_value=0)
           for k in ["ctx", "face", "skel"]}
    out["y"]  = torch.tensor([b["y"]  for b in batch], dtype=torch.float32)
    out["id"] = [b["id"] for b in batch]
    return out

def fast_filter(df, feat_dir, suffix, phase=1):
    dims = {}
    for _, row in df.iterrows():
        p = os.path.join(feat_dir,
            f"{make_key(row['video_id'], row['split'], phase)}_{suffix}.pt")
        if not os.path.exists(p): continue
        try:
            d = torch.load(p, map_location="cpu").shape[-1]
            dims[d] = dims.get(d, 0) + 1
        except: pass
    if not dims: raise ValueError(f"Нет файлов: {feat_dir} suffix={suffix}")
    best = max(dims, key=dims.get)
    idx  = [i for i, (_, row) in enumerate(df.iterrows())
            if os.path.exists(os.path.join(feat_dir,
               f"{make_key(row['video_id'], row['split'], phase)}_{suffix}.pt"))]
    return df.iloc[idx].reset_index(drop=True), best

def load_fold_model(cfg, feat_dim, fold_idx):
    kw  = dict(d=HIDDEN_DIM, n_layers=N_LAYERS, n_heads=N_HEADS, dropout=DROPOUT)
    mod = cfg["modality"]
    if mod == "ctx":    m = ContextModel(d_ctx_in=feat_dim, **kw)
    elif mod == "face": m = FaceModel(d_face_in=feat_dim,   **kw)
    else:               m = SkelModel(d_skel_in=feat_dim,   **kw)
    ckpt = f"/content/{cfg['name']}_fold{fold_idx}.pth"
    if not os.path.exists(ckpt):
        raise FileNotFoundError(f"Нет чекпоинта: {ckpt}")
    m.load_state_dict(torch.load(ckpt, map_location=device))
    return m.to(device).eval()

@torch.no_grad()
def get_probs_df(model, df, cfg, phase=1):
    mod = cfg["modality"]
    ds  = Track3CachedDataset(df, cfg["ctx_dir"], cfg["face_dir"],
                               cfg["skel_dir"], phase=phase)
    dl  = DataLoader(ds, batch_size=32, shuffle=False, collate_fn=collate_fn)
    probs, labels, ids = [], [], []
    for batch in dl:
        probs.extend(torch.sigmoid(model(batch[mod].to(device))).cpu().numpy())
        labels.extend(batch["y"].numpy())
        ids.extend(batch["id"])
    return np.array(probs), np.array(labels), ids

def rank_normalize(p):
    return rankdata(p) / len(p)

def best_thresh_acc(probs, labels):
    best_t, best_acc = 0.5, 0.0
    for t in np.arange(0.05, 0.96, 0.01):
        acc = accuracy_score(labels, (probs >= t).astype(float))
        if acc > best_acc:
            best_acc = acc; best_t = t
    return best_t, best_acc

print("=" * 65)
print("  ШАГ 1 — OOF-предсказания для всех моделей")
print("=" * 65)

oof_store = {}

for cfg in all_model_configs:
    name = cfg["name"]
    mod  = cfg["modality"]
    print(f"\n  {name} ({mod})")

    try:
        _, feat_dim = fast_filter(df_all_labeled, cfg["feat_dir"], mod)
    except Exception as e:
        print(f"    Пропуск (нет фичей): {e}")
        continue

    oof_p = np.full(len(df_all_labeled), np.nan)
    oof_l = np.full(len(df_all_labeled), np.nan)

    for fold_idx, (_, val_idx) in enumerate(cv_splits):
        df_vl_fold = df_all_labeled.iloc[val_idx].reset_index(drop=True)
        try:
            df_vl_f, _ = fast_filter(df_vl_fold, cfg["feat_dir"], mod)
            model_f     = load_fold_model(cfg, feat_dim, fold_idx)
            p_vl, l_vl, ids_vl = get_probs_df(model_f, df_vl_f, cfg)
        except Exception as e:
            print(f"    Fold {fold_idx} пропуск: {e}")
            continue

        id_to_prob  = dict(zip(ids_vl, p_vl))
        id_to_label = dict(zip(ids_vl, l_vl))
        for gi in val_idx:
            vid = int(df_all_labeled.iloc[gi]["video_id"])
            if vid in id_to_prob:
                oof_p[gi] = id_to_prob[vid]
                oof_l[gi] = id_to_label[vid]

    mask  = ~np.isnan(oof_p)
    p_val = oof_p[mask]; l_val = oof_l[mask]

    try:    auc  = roc_auc_score(l_val, p_val)
    except: auc  = float("nan")
    _, top1 = best_thresh_acc(p_val, l_val)

    oof_store[name] = {
        "p":    oof_p,
        "l":    oof_l,
        "mask": mask,
        "auc":  auc,
        "top1": top1,
        "mod":  mod,
        "cfg":  cfg,
        "feat_dim": feat_dim,
    }
    print(f"    OOF top1={top1:.4f} | AUC={auc:.4f} | покрытие={mask.sum()}/{len(df_all_labeled)}")

print("\n" + "=" * 65)
print("  ШАГ 2 — Лучшие одиночные модели по модальностям")
print("=" * 65)

best_single = {}
for name, info in oof_store.items():
    mod = info["mod"]
    if mod not in best_single or info["top1"] > best_single[mod][1]:
        best_single[mod] = (name, info["top1"])

for mod, (name, top1) in best_single.items():
    print(f"  {mod:4s} → {name:12s} top1={top1:.4f}")

best_single_top1 = max(v[1] for v in best_single.values())
print(f"\n  Лучшая одиночная (любая модальность): top1={best_single_top1:.4f}")


print("\n" + "=" * 65)
print("  ШАГ 3 — Перебор всех комбинаций ctx × face × skel")
print("=" * 65)

ctx_models  = [n for n, i in oof_store.items() if i["mod"] == "ctx"]
face_models = [n for n, i in oof_store.items() if i["mod"] == "face"]
skel_models = [n for n, i in oof_store.items() if i["mod"] == "skel"]

weight_grid = [
    (1/3, 1/3, 1/3),
    (0.5, 0.3, 0.2), (0.5, 0.2, 0.3),
    (0.3, 0.5, 0.2), (0.2, 0.5, 0.3),
    (0.3, 0.2, 0.5), (0.2, 0.3, 0.5),
    (0.6, 0.2, 0.2), (0.2, 0.6, 0.2), (0.2, 0.2, 0.6),
    (0.4, 0.4, 0.2), (0.4, 0.2, 0.4), (0.2, 0.4, 0.4),
]

fusion_results = []

for ctx_n, face_n, skel_n in itertools.product(ctx_models, face_models, skel_models):
    mask_all = (
        oof_store[ctx_n]["mask"] &
        oof_store[face_n]["mask"] &
        oof_store[skel_n]["mask"]
    )
    if mask_all.sum() < 10:
        continue

    p_c = oof_store[ctx_n]["p"][mask_all]
    p_f = oof_store[face_n]["p"][mask_all]
    p_s = oof_store[skel_n]["p"][mask_all]
    l   = oof_store[ctx_n]["l"][mask_all]

    p_c_r = rank_normalize(p_c)
    p_f_r = rank_normalize(p_f)
    p_s_r = rank_normalize(p_s)

    best_top1_comb = 0.0
    best_w_comb    = None
    best_t_comb    = 0.5

    for w_c, w_f, w_s in weight_grid:
        fused = w_c * p_c_r + w_f * p_f_r + w_s * p_s_r
        t, acc = best_thresh_acc(fused, l)
        if acc > best_top1_comb:
            best_top1_comb = acc
            best_w_comb    = (w_c, w_f, w_s)
            best_t_comb    = t

    try:
        w_c, w_f, w_s = best_w_comb
        fused_best = w_c * p_c_r + w_f * p_f_r + w_s * p_s_r
        auc_comb   = roc_auc_score(l, fused_best)
    except:
        auc_comb = float("nan")

    fusion_results.append({
        "ctx":    ctx_n,
        "face":   face_n,
        "skel":   skel_n,
        "top1":   best_top1_comb,
        "auc":    auc_comb,
        "weights": best_w_comb,
        "thresh":  best_t_comb,
        "n":       int(mask_all.sum()),
    })

fusion_results.sort(key=lambda x: x["top1"], reverse=True)

print(f"\n  Топ-10 комбинаций:")
print(f"  {'ctx':12s} {'face':12s} {'skel':10s} | {'top1':>6} | {'auc':>6} | weights")
print(f"  {'-'*70}")
for r in fusion_results[:10]:
    w = r["weights"]
    print(f"  {r['ctx']:12s} {r['face']:12s} {r['skel']:10s} | "
          f"{r['top1']:.4f} | {r['auc']:.4f} | "
          f"({w[0]:.2f},{w[1]:.2f},{w[2]:.2f})")


print("\n" + "=" * 65)
print("  ШАГ 4 — Итоговое решение")
print("=" * 65)

best_fusion   = fusion_results[0] if fusion_results else None
use_fusion    = best_fusion is not None and best_fusion["top1"] >= best_single_top1

if use_fusion:
    print(f"\n   Fusion лучше или равен одиночной модели:")
    print(f"    Комбинация : {best_fusion['ctx']} + {best_fusion['face']} + {best_fusion['skel']}")
    print(f"    Веса       : ctx={best_fusion['weights'][0]:.2f} "
          f"face={best_fusion['weights'][1]:.2f} "
          f"skel={best_fusion['weights'][2]:.2f}")
    print(f"    OOF Top-1  : {best_fusion['top1']:.4f} (одиночная: {best_single_top1:.4f})")
    print(f"    OOF AUC    : {best_fusion['auc']:.4f}")

    best_fusion_cfg = {
        "use_fusion":  True,
        "ctx_name":    best_fusion["ctx"],
        "face_name":   best_fusion["face"],
        "skel_name":   best_fusion["skel"],
        "weights":     best_fusion["weights"],
        "top1":        best_fusion["top1"],
    }
else:
    best_mod  = max(best_single, key=lambda m: best_single[m][1])
    best_name = best_single[best_mod][0]
    print(f"\n  ✗ Fusion не даёт прироста. Используем лучшую одиночную:")
    print(f"    Модель     : {best_name} ({best_mod})")
    print(f"    OOF Top-1  : {best_single[best_mod][1]:.4f}")

    best_fusion_cfg = {
        "use_fusion": False,
        "best_name":  best_name,
        "best_mod":   best_mod,
        "top1":       best_single[best_mod][1],
    }

print(f"\n  best_fusion_cfg сохранен → используется в ячейке Phase 2")
```

```python
import torch
from torch.utils.data import DataLoader
from scipy.stats import rankdata
import os, hashlib
import pandas as pd
import numpy as np
from src.ds import Track3CachedDataset
from src.model import ContextModel, FaceModel, SkelModel

device    = "cuda" if torch.cuda.is_available() else "cpu"
BASE_P2   = "/content/drive/MyDrive/miga_features_cache_agcn_p2"
TEST_CSV  = "/content/test_ids.csv"
HIDDEN_DIM = 512; N_LAYERS = 8; N_HEADS = 4; DROPOUT = 0.5

p2_dir_map = {
    "ResNet-18":   {"feat_dir": f"{BASE_P2}/baseline_ctx",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "MC3-18":      {"feat_dir": f"{BASE_P2}/exp_ctx_mc3",
                    "ctx_dir":  f"{BASE_P2}/exp_ctx_mc3",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "R2Plus1D-18": {"feat_dir": f"{BASE_P2}/exp_ctx_r2plus1d",
                    "ctx_dir":  f"{BASE_P2}/exp_ctx_r2plus1d",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "YOLO8":       {"feat_dir": f"{BASE_P2}/baseline_face",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "FaceNet":     {"feat_dir": f"{BASE_P2}/exp_face_facenet",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/exp_face_facenet",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "EmotiEffLib": {"feat_dir": f"{BASE_P2}/exp_face_emotieff",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/exp_face_emotieff",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "2s-AGCN":     {"feat_dir": f"{BASE_P2}/exp_skel_mean_std",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_mean_std"},
    "HRNet":       {"feat_dir": f"{BASE_P2}/exp_skel_hrnet",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_hrnet"},
    "PoseNet":     {"feat_dir": f"{BASE_P2}/exp_skel_posenet",
                    "ctx_dir":  f"{BASE_P2}/baseline_ctx",
                    "face_dir": f"{BASE_P2}/baseline_face",
                    "skel_dir": f"{BASE_P2}/exp_skel_posenet"},
}

name_to_mod = {
    "ResNet-18": "ctx", "MC3-18": "ctx", "R2Plus1D-18": "ctx",
    "YOLO8": "face", "FaceNet": "face", "EmotiEffLib": "face",
    "2s-AGCN": "skel", "HRNet": "skel", "PoseNet": "skel",
}


def make_key(vid, split, phase=1):
    return hashlib.md5(f"{int(vid)}|{split}|{phase}".encode()).hexdigest()[:12]

def collate_fn_test(batch):
    out = {k: torch.nn.utils.rnn.pad_sequence(
               [b[k] for b in batch], batch_first=True, padding_value=0)
           for k in ["ctx", "face", "skel"]}
    out["y"]  = torch.tensor([b["y"]  for b in batch], dtype=torch.float32)
    out["id"] = [b["id"] for b in batch]
    return out

def fast_filter_phase2(df, feat_dir, suffix, phase=2):
    dims = {}
    for _, row in df.iterrows():
        p = os.path.join(feat_dir,
            f"{make_key(row['video_id'], row['split'], phase)}_{suffix}.pt")
        if not os.path.exists(p): continue
        try:
            d = torch.load(p, map_location="cpu").shape[-1]
            dims[d] = dims.get(d, 0) + 1
        except: pass
    if not dims:
        raise ValueError(f"Нет файлов Phase2: {feat_dir} suffix={suffix}")
    best = max(dims, key=dims.get)
    idx  = [i for i, (_, row) in enumerate(df.iterrows())
            if os.path.exists(os.path.join(feat_dir,
               f"{make_key(row['video_id'], row['split'], phase)}_{suffix}.pt"))]
    return df.iloc[idx].reset_index(drop=True), best

def load_fold_model_p2(model_name, feat_dim, fold_idx):
    mod = name_to_mod[model_name]
    kw  = dict(d=HIDDEN_DIM, n_layers=N_LAYERS, n_heads=N_HEADS, dropout=DROPOUT)
    if mod == "ctx":    m = ContextModel(d_ctx_in=feat_dim, **kw)
    elif mod == "face": m = FaceModel(d_face_in=feat_dim,   **kw)
    else:               m = SkelModel(d_skel_in=feat_dim,   **kw)
    ckpt = f"/content/{model_name}_fold{fold_idx}.pth"
    if not os.path.exists(ckpt):
        raise FileNotFoundError(f"Нет чекпоинта: {ckpt}")
    m.load_state_dict(torch.load(ckpt, map_location=device))
    return m.to(device).eval()

@torch.no_grad()
def get_test_probs(model_name, df_test, fold_idx, feat_dim):
    """Прогоняет один фолд одной модели на тест. Возвращает (probs, video_ids)."""
    mod = name_to_mod[model_name]
    dirs = p2_dir_map[model_name]
    model = load_fold_model_p2(model_name, feat_dim, fold_idx)

    ds = Track3CachedDataset(df_test,
                              dirs["ctx_dir"], dirs["face_dir"], dirs["skel_dir"],
                              phase=2)
    dl = DataLoader(ds, batch_size=32, shuffle=False, collate_fn=collate_fn_test)

    probs, ids = [], []
    for batch in dl:
        probs.extend(torch.sigmoid(model(batch[mod].to(device))).cpu().numpy())
        ids.extend(batch["id"])
    return np.array(probs), ids

def rank_normalize(p):
    return rankdata(p) / len(p)


print("=" * 65)
print("  PHASE 2 — ШАГ 1: тестовые ID")
print("=" * 65)

if best_fusion_cfg["use_fusion"]:
    probe_name = best_fusion_cfg["ctx_name"]
else:
    probe_name = best_fusion_cfg["best_name"]

probe_mod  = name_to_mod[probe_name]
probe_dirs = p2_dir_map[probe_name]

if os.path.exists(TEST_CSV):
    df_test = pd.read_csv(TEST_CSV)
    if "split" not in df_test.columns: df_test["split"] = "test"
    if "label" not in df_test.columns: df_test["label"] = -1
    print(f"  Загружен {TEST_CSV}: {len(df_test)} видео")
else:
    feat_dir_probe = probe_dirs["feat_dir"]
    found_keys = set()
    if os.path.exists(feat_dir_probe):
        for fname in os.listdir(feat_dir_probe):
            if fname.endswith(f"_{probe_mod}.pt"):
                found_keys.add(fname.replace(f"_{probe_mod}.pt", ""))

    found_vids = []
    for vid in range(1, 5000):
        k = make_key(vid, "test", 2)
        if k in found_keys:
            found_vids.append(vid)

    df_test = pd.DataFrame({"video_id": found_vids, "split": "test", "label": -1})
    df_test.to_csv(TEST_CSV, index=False)
    print(f"  Обнаружено {len(df_test)} тестовых видео → сохранено в {TEST_CSV}")


print("\n" + "=" * 65)
print(f"  PHASE 2 — ШАГ 2: soft ensemble {N_FOLDS} фолдов")
print("=" * 65)

if best_fusion_cfg["use_fusion"]:
    models_needed = [
        best_fusion_cfg["ctx_name"],
        best_fusion_cfg["face_name"],
        best_fusion_cfg["skel_name"],
    ]
else:
    models_needed = [best_fusion_cfg["best_name"]]

print(f"  Модели для прогона: {models_needed}")

te_store = {}

for model_name in models_needed:
    mod  = name_to_mod[model_name]
    dirs = p2_dir_map[model_name]
    print(f"\n  {model_name} ({mod})")

    try:
        df_te_filtered, feat_dim = fast_filter_phase2(
            df_test, dirs["feat_dir"], mod, phase=2)
    except Exception as e:
        print(f"    Ошибка фильтрации: {e}")
        te_store[model_name] = {"p": np.array([]), "ids": []}
        continue

    all_fold_probs = []
    ids_te = None

    for fold_idx in range(N_FOLDS):
        try:
            p_te, ids_te = get_test_probs(model_name, df_te_filtered, fold_idx, feat_dim)
            all_fold_probs.append(p_te)
            print(f"    Fold {fold_idx}: n={len(p_te)} "
                  f"probs=[{p_te.min():.3f}..{p_te.max():.3f}]")
        except Exception as e:
            print(f"    Fold {fold_idx} пропуск: {e}")

    if not all_fold_probs:
        print(f"    Нет чекпоинтов для {model_name}")
        te_store[model_name] = {"p": np.array([]), "ids": []}
        continue

    p_mean = np.mean(all_fold_probs, axis=0)
    te_store[model_name] = {"p": p_mean, "ids": ids_te}
    print(f"    Ensemble: [{p_mean.min():.3f}..{p_mean.max():.3f}]")


print("\n" + "=" * 65)
print("  PHASE 2 — ШАГ 3: формирование submission.csv")
print("=" * 65)

if best_fusion_cfg["use_fusion"]:
    ctx_n  = best_fusion_cfg["ctx_name"]
    face_n = best_fusion_cfg["face_name"]
    skel_n = best_fusion_cfg["skel_name"]
    w_c, w_f, w_s = best_fusion_cfg["weights"]

    ids_c = set(te_store[ctx_n]["ids"])
    ids_f = set(te_store[face_n]["ids"])
    ids_s = set(te_store[skel_n]["ids"])
    common_ids = sorted(ids_c & ids_f & ids_s)
    print(f"  Fusion: {ctx_n} + {face_n} + {skel_n}")
    print(f"  Веса  : ctx={w_c:.2f} face={w_f:.2f} skel={w_s:.2f}")
    print(f"  Общих тестовых ID: {len(common_ids)}")

    def probs_for_ids(model_name, ids):
        d   = te_store[model_name]
        idx = {vid: i for i, vid in enumerate(d["ids"])}
        return np.array([d["p"][idx[vid]] for vid in ids])

    p_c_te = probs_for_ids(ctx_n,  common_ids)
    p_f_te = probs_for_ids(face_n, common_ids)
    p_s_te = probs_for_ids(skel_n, common_ids)

    p_c_r = rank_normalize(p_c_te)
    p_f_r = rank_normalize(p_f_te)
    p_s_r = rank_normalize(p_s_te)

    fused  = w_c * p_c_r + w_f * p_f_r + w_s * p_s_r
    thresh = float(np.median(fused))
    preds  = (fused >= thresh).astype(int)

    submission = pd.DataFrame({"video_id": common_ids, "label": preds})

    all_test_ids = set(df_test["video_id"].tolist())
    missing = sorted(all_test_ids - set(common_ids))

else:
    best_name = best_fusion_cfg["best_name"]
    print(f"  Одиночная модель: {best_name}")
    p_te  = te_store[best_name]["p"]
    ids_te = te_store[best_name]["ids"]

    p_r    = rank_normalize(p_te)
    thresh = float(np.median(p_r))
    preds  = (p_r >= thresh).astype(int)

    submission = pd.DataFrame({"video_id": ids_te, "label": preds})
    all_test_ids = set(df_test["video_id"].tolist())
    missing = sorted(all_test_ids - set(ids_te))

if missing:
    print(f"  {len(missing)} видео без фичей → label=1 (prior)")
    submission = pd.concat([
        submission,
        pd.DataFrame({"video_id": missing, "label": 1})
    ], ignore_index=True)

submission = submission.sort_values("video_id").reset_index(drop=True)

print(f"\n  Итого в сабмите : {len(submission)}")
print(f"  label=1 (win)   : {int(submission['label'].sum())}")
print(f"  label=0 (loss)  : {int((submission['label'] == 0).sum())}")
print(f"\n  Первые 10 строк:")
print(submission.head(10).to_string(index=False))

submission.to_csv("/content/submission.csv", index=False)
print("\n  Сохранено: /content/submission.csv")

print("\n" + "=" * 65)
print("  ИТОГ")
print("=" * 65)
if best_fusion_cfg["use_fusion"]:
    print(f"  Стратегия  : Late Fusion ({best_fusion_cfg['ctx_name']} + "
          f"{best_fusion_cfg['face_name']} + {best_fusion_cfg['skel_name']})")
    print(f"  OOF Top-1  : {best_fusion_cfg['top1']:.4f}")
else:
    print(f"  Стратегия  : Одиночная модель ({best_fusion_cfg['best_name']})")
    print(f"  OOF Top-1  : {best_fusion_cfg['top1']:.4f}")
print(f"  Порог      : {thresh:.4f} (медиана rank-score на тесте)")
```
